#Map

```python
intlist = [1,2,3,4,5,6,7,8,9]
```

```python
squarelist = []
```

```python
for x in intlist:
    squarelist.append(pow(x,2))

print(squarelist)
```

```python
def square_me(x):
    return x * x
```

```python
squarelist = map(square_me,intlist)
```

```python
list(squarelist)
```

```python
squarelist = map(lambda x: x*x, intlist)
list(squarelist)
```

#Filter

```python
evenints = filter(lambda x:x % 2 == 0, intlist)
list(evenints)
```
